## Testskript


Das Testskript ruft Ihre Funktionen mit Gleichungsystemen der Dimension $n = 15, 25, 50, 100$ und Toleranz $\mathtt{tol} = 10^{-4}$ auf und vergleicht die Ergebnisse.
Da das etwas dauern kann, können Sie zum Testen kleinere Parameter wählen.

```python
Ns = [15, 25, 50, 100]
#Ns = [10, 15, 25, 50]
```

```python
def build_A(N):
    """
        Erstelle eine Laplace-Matrix der Größe (N,N) als Testsystem
    """
    f = (N+1)**2
    D = np.full(N, 2*f)
    E = np.full(N-1, f)
    A = np.diag(D) - np.diag(E, k=1) - np.diag(E, k=-1)
    return A

def build_b(N):
    """
        Erzeuge die rechte Seite des Gleichungssystems
    """
    x = np.linspace(0, np.pi, N+2)
    y = (np.pi*np.pi) * np.sin(x)[1:-1]
    return y

max_iter = 50_000
tol = 1e-4
fig = plt.figure(figsize=(10,5))
px = 0
for (i, N) in enumerate(Ns):
    #  Grafik-Initialisierung
    plt.subplot(2, 2, i+1)
    plt.title(f"{N=}", fontsize=14)
    
    #  Initialisierung des linearen Gleichungssystems
    A = build_A(N)
    b = build_b(N)
    x = np.linspace(0.0, 1.0, N+2)[1:-1]

    print(f"{N=}")

    #  1.Verfahren:  numpy-Löser
    start = time.perf_counter()
    reference_solution = np.linalg.solve(A, b)
    numpy_time = time.perf_counter() - start
    numpy_res = norm(A @ reference_solution - b, np.inf)
    print(f"numpy.linalg.solve:\t\ttime = {numpy_time:.2f}\tres = {numpy_res:.4e}")
    
    #  2.Verfahren:  Gauss-Seidel
    start = time.perf_counter()
    gauss_solution, gauss_iter = gauss_seidel(A, b, max_iter=max_iter, tol=tol)
    gauss_time = time.perf_counter() - start
    gauss_res  = norm(A @ gauss_solution - b, np.inf)
    print(f"Gauss:\titerations = {gauss_iter}\ttime = {gauss_time:.2f}s\tres = {gauss_res:.4e}")
    plt.plot(x, reference_solution-gauss_solution, "g-", label=f"Gauss-Seidel  k={gauss_iter}")
    
    #  3.Verfahren:  Jacobi
    start = time.perf_counter()
    jacobi_solution, jacobi_iter = jacobi(A, b, max_iter=max_iter, tol=tol)
    jacobi_time = time.perf_counter() - start
    jacobi_res  = norm(A @ jacobi_solution - b, np.inf)
    print(f"Jacobi:\titerations = {jacobi_iter}\ttime = {jacobi_time:.2f}s\tres = {jacobi_res:.4e}")
    plt.plot(x, reference_solution-jacobi_solution, "b-", label=f"Jacobi  k={jacobi_iter}")
    plt.grid(True)
    plt.legend(loc=1)
plt.tight_layout()
plt.show()
```
